# Dominick's Data

## Libraries

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent ))
```

```python
from src.dominick import DominickDataLoader, DominickDataProcessor, DominickDataSaver
```

## Loader

```python
loader = DominickDataLoader()
df_upc_beer = loader.load("upcber.csv")
df_move_beer = loader.load("wber.csv")
```

## Join

```python
dataprocessor = DominickDataProcessor()
df_upc_move_beer = dataprocessor.join(df_upc_beer, df_move_beer, on="UPC")
df_upc_move_beer.head()
```

## Drop HEX

```python
df_upc_move_beer = df_upc_move_beer.drop(columns=["PROFIT_HEX", "PRICE_HEX"])
df_upc_move_beer.head()
```

## Renaming

```python
rename_dict = {
    "COM_CODE": "category_code",
    "UPC": "upc_code",
    "DESCRIP": "product_description",
    "SIZE": "pack_size_text",
    "CASE": "units_per_case",
    "NITEM": "product_item_code",
    "STORE": "store_code",
    "WEEK": "week_id",
    "MOVE": "units_sold",
    "QTY": "units_per_deal",
    "PRICE": "total_price",
    "SALE": "promo_flag",
    "PROFIT": "gross_margin_pct",
    "OK": "exclude_flag",
}

df_upc_move_beer = dataprocessor.rename(df_upc_move_beer, rename_dict)
df_upc_move_beer.head()
```

# Dominick's Data

> KILTS Center for Marketing: Dominick's Data Manual  
> We do not use flagged data in our analysis. Therefore, we exclude it: `exclude_flag != 0`

```python
df_upc_move_beer_flagged = df_upc_move_beer[df_upc_move_beer["exclude_flag"] != 0].drop(columns=["exclude_flag"])
df_upc_move_beer_flagged.head()
```

## New Ratios

```python
# We add the financial ratios to the DataFrame.
df_new_ratios = dataprocessor.add_ratios(df_upc_move_beer_flagged)
df_new_ratios.head()
```

# New Descriptions

```python
# We add the text normalization from product_description.
df_new_descriptions = dataprocessor.add_text_normalization(df_new_ratios)
df_new_descriptions.head()
```

## Elasticity features

```python
# We add the elasticity features to the DataFrame.
df_elasticity_features = dataprocessor.add_elasticity_features(df_new_descriptions)
df_elasticity_features.head()
```

## Save CSV

```python
saver = DominickDataSaver()
saver.save("dominick_features.csv", df_elasticity_features)
```
